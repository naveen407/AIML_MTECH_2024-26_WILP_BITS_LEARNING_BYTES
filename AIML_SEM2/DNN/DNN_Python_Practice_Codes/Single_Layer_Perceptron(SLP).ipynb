```python
import numpy as np

#Define a class SingleLayerPerceptron to encapsulate the functionality of
#perceptron.

class SingleLayerPerceptron:
    def __init__(self, input_size):
        # Initialize weights and bias
        self.weights = np.zeros(input_size)
        self.bias = 0
#The __init__ method is the constructor, which initializes the
#perceptron's weights and bias. In this example, we start with zero weights
#and a bias of zero. The number of weights is equal to the input_size. self is
#the instance of the class 'SingleLayerPerceptron'

    def predict(self, inputs):
        # Calculate the weighted sum and apply the step function
        weighted_sum = np.dot(inputs, self.weights) + self.bias
        return 1 if weighted_sum > 0 else 0

#The predict method takes an input vector (inputs) and calculates the weighted
#sum of the inputs using the perceptron's weights and bias. It then applies a
#simple step function: if the weighted sum is greater than 0, it returns 1;
#otherwise, it returns 0. This is a basic thresholding operation that simulates
#the activation function of a perceptron.

    def train(self, inputs, labels, learning_rate=0.1, epochs=100):
        for epoch in range(epochs):
            for i in range(len(inputs)):
                # Make a prediction
                prediction = self.predict(inputs[i])

                # Update weights and bias based on prediction error
                self.weights += learning_rate * (labels[i] - prediction) * inputs[i]
                self.bias += learning_rate * (labels[i] - prediction)

#The train method is used to train the perceptron. It iterates over the dataset
#for a specified number of epochs. Within each epoch, it goes through each input
#and updates the weights and bias based on the prediction error.
#The update rule is a simple gradient descent-like approach where the weights
#and bias are adjusted in the direction that reduces the prediction error.

# Example usage
# Assuming a binary classification problem with two features
inputs = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
labels = np.array([0, 0, 0, 1])

# Create and train the perceptron
input_size = len(inputs[0])
perceptron = SingleLayerPerceptron(input_size)
perceptron.train(inputs, labels)

# Make predictions
predictions = [perceptron.predict(x) for x in inputs]

# Print predictions
print("Inputs:", inputs)
print("Predictions:", predictions)

#In this example usage section, we create a dataset (inputs and labels) for
#a binary classification problem (in this case, the AND problem).
#We create an instance of the SingleLayerPerceptron class, train it on the
#dataset, and then make predictions on the same dataset. Finally, we print the
#inputs and the corresponding predictions.
```

```python
import matplotlib.pyplot as plt

# Example usage
inputs = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
labels = np.array([0, 0, 0, 1])

input_size = len(inputs[0])
perceptron = SingleLayerPerceptron(input_size)

# Visualize the data points
plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.scatter(inputs[:, 0], inputs[:, 1], c=labels, cmap='viridis', marker='o', s=200)
plt.title('Data Points')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')

# Train the perceptron
perceptron.train(inputs, labels)

# Visualize the decision boundary
plt.subplot(1, 2, 2)
plt.scatter(inputs[:, 0], inputs[:, 1], c=labels, cmap='viridis', marker='o', s=200, label='Data Points')
plt.title('Decision Boundary')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')

# Plot the  decision boundary (after training)
x_vals = np.linspace(-0.1, 1.1, 100)
y_vals = -(perceptron.weights[0] * x_vals + perceptron.bias) / perceptron.weights[1]
plt.plot(x_vals, y_vals, '--', label='Decision Boundary')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np

class SingleLayerPerceptron:
    def __init__(self, input_size):
        self.weights = np.zeros(input_size)
        self.bias = 0

    def predict(self, inputs):
        weighted_sum = np.dot(inputs, self.weights) + self.bias
        return 1 if weighted_sum > 0 else 0

    def train(self, inputs, labels, learning_rate=0.1, epochs=100):
        for epoch in range(epochs):
            for i in range(len(inputs)):
                prediction = self.predict(inputs[i])
                self.weights += learning_rate * (labels[i] - prediction) * inputs[i]
                self.bias += learning_rate * (labels[i] - prediction)

            predictions = [self.predict(x) for x in inputs]
            accuracy = np.mean(predictions == labels)
            print(f"Epoch {epoch + 1}/{epochs}, Accuracy: {accuracy:.2%}, Predictions: {predictions}")

# Example usage
inputs = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
labels = np.array([0, 0, 0, 1])

input_size = len(inputs[0])
perceptron = SingleLayerPerceptron(input_size)
perceptron.train(inputs, labels)
```
