Purpose:
Build decoder-only Transformer-based language model from scratch (like GPT), using PyTorch. It supports tokenization, training, text generation, checkpointing, and inference.

Main Components
1. Tokenization (Tokenizer)
Character-level tokenizer.
Supports digits (0-9), lowercase letters (a-z), space ( ), period (.), and a padding token (<pad>).
Maps characters to unique integer tokens and vice versa.
2. Transformer Architecture
TokenEmbedding
Maps each input token ID to a dense vector (d_model dimension).

PositionalEncoding
Adds sinusoidal positional information to token embeddings.
Helps the model understand the order of tokens.

MaskedSelfAttention
Computes scaled dot-product attention.
Uses a mask to prevent tokens from attending to future tokens.

MaskedMultiHeadedSelfAttention
Runs multiple attention heads in parallel.
Concatenates outputs and applies a linear projection.

FeedForward
Standard 2-layer feedforward neural net with ReLU.

DecoderLayer
One layer of transformer decoder.
LayerNorm → MultiHeadAttention + residual → LayerNorm → FeedForward + residual.

DecoderStack
Stack of DecoderLayers.
Processes input embeddings through all decoder layers.

LMHead
Linear layer mapping from embedding dimension to vocabulary size.
Outputs logits over the vocabulary.

LanguageModel
Full decoder-only transformer model (like GPT).
TokenEmbedding + PositionalEncoding → DecoderStack → LMHead.

AutoregressiveWrapper
Adds autoregressive behavior.
Implements teacher forcing during training.
Provides next_token_probabilities() for inference.

3. Training Pipeline
Trainer
Prepares batches with padding and masking.
Uses cross-entropy loss (teacher forcing).
Applies gradient clipping.
Logs average loss per epoch.

4. Text Generation
Generator
Autoregressively samples one token at a time.
Starts from a prompt or padding.
Stops on <eos> or max_tokens_to_generate.


5. Utility Functions
get_device(): Detects CUDA/GPU.
pad_left(): Pads sequences to fixed length from the left.
create_training_sequences(): Generates sliding windows of max_length + 1 tokens (context + target).
tokenize_and_pad_training_data(): Tokenizes string and prepends padding.


6. Runner
Orchestrates the entire workflow:
Initializes tokenizer and model.
Prepares training data.
Trains the model for 120 epochs.
Plots training loss (log scale).
Saves model checkpoint.
Generates 400 characters from prompt "cats".
